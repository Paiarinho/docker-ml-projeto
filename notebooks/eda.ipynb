```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sqlalchemy import create_engine
```

```python
path = "data/heat_exchanger.db"
```

```python
# Passo 1: Leitura do dataset
engine = create_engine(f"sqlite:///{path}")

df = pd.read_sql("SELECT * FROM heat_exchanger", engine)
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['day_index'] = (df['timestamp'] - df['timestamp'].min()).dt.days
df.head()
```

```python
# Passo 2: Visualizacao de dados
plt.figure(figsize=(12, 6))
plt.plot(df['day_index'], df['heat_efficiency'])
plt.xlabel('Day index')
plt.ylabel('Heat Efficiency')
plt.grid()
plt.show()
```

```python
# Parte 3: Predicao da eficiencia termica
y = df['heat_efficiency'].values
x = df['day_index'].values

# Encontrar coeficientes da reta utilizando numpy
coef = np.polyfit(x, y, 1)
pred = np.polyval(coef, x)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df['day_index'], df['heat_efficiency'], marker='.', label="Dados Reais")
plt.plot(x, pred, label="Dados Calculados")

plt.xlabel('Day index')
plt.ylabel('Heat Efficiency')

plt.legend(loc='upper right')

plt.grid()
plt.show()
```

```python
# Calculo do r2

residuos = y - pred
ss_res = np.sum(residuos**2)
ss_tot = np.sum((y - np.mean(y))**2)

r2 = 1 - (ss_res / ss_tot)

print(f"R²: {r2:.4f}")
```
